```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import KNeighborsClassifier

# Twitter data
tweets = [
    "I love this phone", "Amazing product", "Very happy today",
    "This is excellent", "I hate this phone", "Worst product",
    "Very bad experience", "I am disappointed"
]

# 1 = Positive, 0 = Negative
y = [1, 1, 1, 1, 0, 0, 0, 0]

# Convert text to numbers
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(tweets)

# KNN classifier
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

# Test tweet
test = ["I am very happy with this product"]
test_X = vectorizer.transform(test)

prediction = model.predict(test_X)

print("Tweet:", test[0])
print("Sentiment:", "Positive" if prediction[0] == 1 else "Negative")
```
