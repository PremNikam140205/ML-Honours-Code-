```python
import numpy as np

def hebb_train(X, y):
    w = np.zeros(X.shape[1])
    b = 0

    for x, target in zip(X, y):
        w += x * target
        b += target

    return w, b

def predict(X, w, b):
    return np.where(np.dot(X, w) + b >= 0, 1, 0)

# AND Gate
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([-1, -1, -1, 1])

w, b = hebb_train(X, y)
print("AND:", predict(X, w, b))

# OR Gate
y = np.array([-1, 1, 1, 1])

w, b = hebb_train(X, y)
print("OR :", predict(X, w, b))

# NOT Gate
X = np.array([[0], [1]])
y = np.array([1, -1])

w, b = hebb_train(X, y)
print("NOT:", predict(X, w, b))
```
